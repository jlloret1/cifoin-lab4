```python
# MMI

import gdsfactory as gf

layout = gf.Component()

# Parámetros obtenidos del LAB2
mmi_width = 6.6
mmi_length = 34.2180      # 34.5180 - 0.3000

waveguide_width = 1.0
access_length = 10.0

dy = 0.1

bend_dx = 30.0
bend_dy = 40.0

# Posiciones de los puertos
lower_port_y = mmi_width * (-1/6) - dy
upper_port_y = mmi_width * ( 1/6) + dy

# Bloque multimodo central
mmi_region = layout.add_ref(
    gf.components.rectangle(
        size=(mmi_length, mmi_width),
        layer="WG"
    )
)

# Centrar el bloque
mmi_region.dmove(
    (-mmi_length/2, -mmi_width/2)
)

# Sección transversal
xs = gf.cross_section.strip(
    width=waveguide_width,
    layer="WG"
)

# Guías de acceso
access_waveguide = gf.components.straight(
    length=access_length,
    cross_section=xs
)

# Entrada superior izquierda (o2)
left_upper_access = layout.add_ref(access_waveguide)
left_upper_access.drotate(180)
left_upper_access.dmove((-mmi_length/2, upper_port_y))

# Entrada inferior izquierda (o1)
left_lower_access = layout.add_ref(access_waveguide)
left_lower_access.drotate(180)
left_lower_access.dmove((-mmi_length/2, lower_port_y))

# Salida superior derecha (o3)
right_upper_access = layout.add_ref(access_waveguide)
right_upper_access.dmove((mmi_length/2, upper_port_y))

# Salida inferior derecha (o4)
right_lower_access = layout.add_ref(access_waveguide)
right_lower_access.dmove((mmi_length/2, lower_port_y))

# Curva superior izquierda
left_upper_bend = layout.add_ref(
    gf.components.bend_s(
        size=(bend_dx, bend_dy),
        cross_section=xs
    )
)
left_upper_bend.connect(
    "o1",
    left_lower_access.ports["o2"]
)

# Curva inferior izquierda
left_lower_bend = layout.add_ref(
    gf.components.bend_s(
        size=(bend_dx, bend_dy),
        cross_section=xs
    )
).mirror()

left_lower_bend.connect(
    "o1",
    left_upper_access.ports["o2"]
)

# Curva superior derecha
right_upper_bend = layout.add_ref(
    gf.components.bend_s(
        size=(bend_dx, bend_dy),
        cross_section=xs
    )
)

right_upper_bend.connect(
    "o1",
    right_upper_access.ports["o2"]
)

# Curva inferior derecha
right_lower_bend = layout.add_ref(
    gf.components.bend_s(
        size=(bend_dx, bend_dy),
        cross_section=xs
    )
).mirror()

right_lower_bend.connect(
    "o1",
    right_lower_access.ports["o2"]
)

# Puertos
layout.add_port(
    "o1",
    port=left_lower_bend.ports["o2"]
)

layout.add_port(
    "o2",
    port=left_upper_bend.ports["o2"]
)

layout.add_port(
    "o3",
    port=right_upper_bend.ports["o2"]
)

layout.add_port(
    "o4",
    port=right_lower_bend.ports["o2"]
)

layout.draw_ports()

layout.plot()
```

```python
# RING RESONATOR

gf.clear_cache()

layout = gf.Component()

# Parámetros
mmi_width = 6.6
mmi_length = 34.2180

waveguide_width = 1.0
access_length = 10.0

dy = 0.1

ring_radius = 50.0

xs = gf.cross_section.strip(
    width=waveguide_width,
    layer="WG"
)

# MMI central


# Posiciones de los puertos
lower_port_y = mmi_width * (-1/6) - dy
upper_port_y = mmi_width * ( 1/6) + dy

# Bloque central
mmi_region = layout.add_ref(
    gf.components.rectangle(
        size=(mmi_length, mmi_width),
        layer="WG"
    )
)

# Centrar bloque
mmi_region.dmove(
    (-mmi_length / 2, -mmi_width / 2)
)

# Guías de acceso
access_waveguide = gf.components.straight(
    length=access_length,
    cross_section=xs
)

# Entrada superior izquierda (o2)
left_upper_access = layout.add_ref(access_waveguide)
left_upper_access.drotate(180)
left_upper_access.dmove((-mmi_length / 2, upper_port_y))

# Entrada inferior izquierda (o1)
left_lower_access = layout.add_ref(access_waveguide)
left_lower_access.drotate(180)
left_lower_access.dmove((-mmi_length / 2, lower_port_y))

# Salida superior derecha (o3)
right_upper_access = layout.add_ref(access_waveguide)
right_upper_access.dmove((mmi_length / 2, upper_port_y))

# Salida inferior derecha (o4)
right_lower_access = layout.add_ref(access_waveguide)
right_lower_access.dmove((mmi_length / 2, lower_port_y))


# Lado derecho del anillo

right_bend_1 = layout.add_ref(
    gf.components.bend_circular(
        angle=90,
        radius=ring_radius,
        cross_section=xs
    )
)

right_bend_1.connect(
    "o1",
    right_upper_access.ports["o2"]
)

right_bend_2 = layout.add_ref(
    gf.components.bend_circular(
        angle=90,
        radius=ring_radius,
        cross_section=xs
    )
)

right_bend_2.connect(
    "o1",
    right_bend_1.ports["o2"]
)


# Lado izquierdo del anillo

left_bend_1 = layout.add_ref(
    gf.components.bend_circular(
        angle=-90,
        radius=ring_radius,
        cross_section=xs
    )
)

left_bend_1.connect(
    "o1",
    left_upper_access.ports["o2"]
)

left_bend_2 = layout.add_ref(
    gf.components.bend_circular(
        angle=-90,
        radius=ring_radius,
        cross_section=xs
    )
)

left_bend_2.connect(
    "o1",
    left_bend_1.ports["o2"]
)

# Separación entre puertos superiores

port_separation = abs(
    left_upper_access.ports["o2"].dy
    - right_upper_access.ports["o2"].dy
)

# Recta superior del anillo

upper_waveguide = layout.add_ref(
    gf.components.straight(
        length=port_separation,
        cross_section=xs
    )
)

upper_waveguide.connect(
    "o1",
    right_bend_2.ports["o2"]
)

gf.routing.route_single(
    component=layout,
    port1=upper_waveguide.ports["o2"],
    port2=left_bend_2.ports["o2"],
    cross_section=xs
)


# Puertos externos

layout.add_port(
    name="o1",
    port=left_lower_access.ports["o2"]
)

layout.add_port(
    name="o2",
    port=left_upper_access.ports["o2"]
)

layout.add_port(
    name="o3",
    port=right_upper_access.ports["o2"]
)

layout.add_port(
    name="o4",
    port=right_lower_access.ports["o2"]
)

layout.draw_ports()

layout.plot()
```

### Mantra: 
Once that we have a working 'new' component, we shall convert it into a Cell. This will allow us to have a hierarchical design

```python
# 1. We define the cell as a function with it's corresponding parameters (and defaults)
gf.clear_cache()

@gf.cell
def die(dieL = 5000, dieW = 5000, border = 250, layer_box = "FLOORPLAN"):
    # Die specifications (Chip)
    box = gf.Component()
    obox = box.add_ref(gf.components.rectangle(size=(dieW,dieL),layer=layer_box))
    ibox = box.add_ref(gf.components.rectangle(size=(dieW-border*2,dieL-border*2),layer=layer_box)).dmovex(border).dmovey(border)
    box = gf.boolean(A=obox, B=ibox, operation="A-B", layer=layer_box)
    # Adding ports to a component 
    box.add_port(name="block@org", center=[border,border], width=1, orientation=0, layer=layer_box)
    box.draw_ports()
    return box

# 2. We instantiate references of each cell. We can also name it individually to avoid problems
wafer = gf.Component()
dieW1 = 5000
dieW2 = 10000

c1 = wafer.add_ref(die(dieW = dieW))
c2 = wafer.add_ref(die(dieW = dieW))
c3 = wafer.add_ref(die(dieW = 12000))

c2.dmovex(dieW1)
c3.dmovey(-5000)
wafer.show()
wafer.plot()
```

```python
#DC CELL
import gdsfactory as gf

gf.clear_cache()

@gf.cell
def directional_coupler(
    length=47.014,
    gap=0.6,
    dx=30.0,
    dy=40.0,
    width=1.2,
):

    # Creamos el componente
    c = gf.Component()

    # Definimos la sección transversal tipo Strip
    xs = gf.cross_section.strip(
        width=width,
        layer="WG"
    )

    # Región de acoplamiento
    coupling_region = c.add_ref(
        gf.components.coupler_straight(
            length=length,
            gap=gap,
            cross_section=xs
        )
    )

    # Curva superior izquierda
    left_top_bend = c.add_ref(
        gf.components.bend_s(
            size=(dx, dy),
            cross_section=xs
        )
    )
    left_top_bend.connect("o1", coupling_region.ports["o1"])

    # Curva inferior izquierda
    left_bottom_bend = c.add_ref(
        gf.components.bend_s(
            size=(dx, dy),
            cross_section=xs
        )
    ).mirror()

    left_bottom_bend.connect(
        "o1",
        coupling_region.ports["o2"]
    )

    # Curva superior derecha
    right_top_bend = c.add_ref(
        gf.components.bend_s(
            size=(dx, dy),
            cross_section=xs
        )
    )

    right_top_bend.connect(
        "o1",
        coupling_region.ports["o3"]
    )

    # Curva inferior derecha
    right_bottom_bend = c.add_ref(
        gf.components.bend_s(
            size=(dx, dy),
            cross_section=xs
        )
    ).mirror()

    right_bottom_bend.connect(
        "o1",
        coupling_region.ports["o4"]
    )

    # Puertos
    c.add_port("o1", port=left_bottom_bend.ports["o2"])
    c.add_port("o2", port=left_top_bend.ports["o2"])
    c.add_port("o3", port=right_top_bend.ports["o2"])
    c.add_port("o4", port=right_bottom_bend.ports["o2"])

    c.draw_ports()

    return c


# Instanciación de la celda

dc1 = directional_coupler()
dc2 = directional_coupler(length=20.0) #Comprobamos que el parámetro se puede modificar sin problemas

dc1.plot()
dc2.plot()
```

```python
#MMI CELL


import gdsfactory as gf

gf.clear_cache()

@gf.cell
def mmi_2x2(
    W_MMI=6.6,
    L_MMI=34.2180,
    W_wg=1.0,
    L_wg=10.0,
    dy=0.1,
    dx_bend=30.0,
    dy_bend=40.0,
):
    c = gf.Component()

    #puertos
    pos_inf = W_MMI * (-1/6) + (-dy)
    pos_sup = W_MMI * ( 1/6) + ( dy)

 
    mmi_core = c.add_ref(
        gf.components.rectangle(size=(L_MMI, W_MMI), layer="WG"))
    mmi_core.dmove((-L_MMI / 2, -W_MMI / 2))

    # Guías 
    xs = gf.cross_section.strip(width=W_wg, layer='WG', radius_min=0, radius=0)
    wg = gf.components.straight(length=L_wg, cross_section=xs)

    # Entrada superior izquierda
    t_o2 = c.add_ref(wg)
    t_o2.drotate(180)
    t_o2.dmove((-L_MMI / 2, pos_sup))

    # Entrada inferior izquierda 
    t_o1 = c.add_ref(wg)
    t_o1.drotate(180)
    t_o1.dmove((-L_MMI / 2, pos_inf))

    # Salida superior derecha
    t_o3 = c.add_ref(wg)
    t_o3.dmove((L_MMI / 2, pos_sup))

    # Salida inferior derecha 
    t_o4 = c.add_ref(wg)
    t_o4.dmove((L_MMI / 2, pos_inf))

  
    bs_sup_izq = c.add_ref(gf.components.bend_s(size=(dx_bend, dy_bend), cross_section=xs))
    bs_sup_izq.connect("o1", t_o1.ports["o2"])

    bs_inf_izq = c.add_ref(gf.components.bend_s(size=(dx_bend, dy_bend), cross_section=xs)).mirror()
    bs_inf_izq.connect("o1", t_o2.ports["o2"])

    bs_sup_der = c.add_ref(gf.components.bend_s(size=(dx_bend, dy_bend), cross_section=xs))
    bs_sup_der.connect("o1", t_o3.ports["o2"])

    bs_inf_der = c.add_ref(gf.components.bend_s(size=(dx_bend, dy_bend), cross_section=xs)).mirror()
    bs_inf_der.connect("o1", t_o4.ports["o2"])

    # Puertos
    c.add_port(name="o1", port=bs_inf_izq.ports["o2"])
    c.add_port(name="o2", port=bs_sup_izq.ports["o2"])
    c.add_port(name="o3", port=bs_sup_der.ports["o2"])
    c.add_port(name="o4", port=bs_inf_der.ports["o2"])

    c.draw_ports()
    return c


# Instanciación
chip_mmi2x2_bend = gf.Component()

mmi1_b = chip_mmi2x2_bend.add_ref(mmi_2x2(L_MMI=50))
mmi2_b = chip_mmi2x2_bend.add_ref(mmi_2x2())
mmi2_b.dmovey(100)

chip_mmi2x2_bend.plot()
```

```python
@gf.cell
def mzi(
    delta_length=100.0,
    L_brazo=400.0,
):
    c  = gf.Component()
    xs = gf.cross_section.strip(width=1.0, layer="WG", radius_min=0, radius=0)  

    MMI_izq = c.add_ref(mmi_2x2())

    brazo_sup = c.add_ref(gf.components.straight(length=L_brazo + delta_length, cross_section=xs))
    brazo_inf = c.add_ref(gf.components.straight(length=L_brazo, cross_section=xs))

    brazo_sup.connect("o1", MMI_izq.ports["o3"])
    brazo_inf.connect("o1", MMI_izq.ports["o4"])

    brazo_inf_fill = c.add_ref(gf.components.straight(length=delta_length, cross_section=xs))
    brazo_inf_fill.connect("o1", brazo_inf.ports["o2"])

    MMI_der = c.add_ref(mmi_2x2())
    MMI_der.dmirror()
    MMI_der.connect("o3", brazo_sup.ports["o2"])

    c.add_port(name="o1", port=MMI_izq.ports["o1"])
    c.add_port(name="o2", port=MMI_izq.ports["o2"])
    c.add_port(name="o3", port=MMI_der.ports["o1"])
    c.add_port(name="o4", port=MMI_der.ports["o2"])

    c.draw_ports()
    return c


chip_MZI = gf.Component()

mzi1 = chip_MZI.add_ref(mzi( L_brazo=400.0,delta_length=500.0))
mzi2 = chip_MZI.add_ref(mzi())
mzi2.dmovey(120)

chip_MZI.plot()
```

```python
#RING RESONATOR CELL
gf.clear_cache()

@gf.cell
def ring_resonator_r(
    W_MMI   = 6.6,
    L_MMI   = 34.218,
    W_wg    = 1.0,
    L_wg    = 10.0,
    dy      = 0.1,
    radio   = 60.0, 
    L_extra = 0.0,
):
    c  = gf.Component()
    xs = gf.cross_section.strip(width=W_wg, layer="WG")
    
    # MMI 
    mmi = c.add_ref(mmi2x2(W_MMI=W_MMI, L_MMI=L_MMI, W_wg=W_wg, L_wg=L_wg, dy=dy))

    # Lado derecho
    bend_der_1 = c.add_ref(gf.components.bend_euler(angle=180, radius=radio, cross_section=xs))
    bend_der_1.connect("o1", mmi.ports["o3"])

    # Lado izquierdo
    bend_izq_1 = c.add_ref(gf.components.bend_euler(angle=-180, radius=radio, cross_section=xs))
    bend_izq_1.connect("o1", mmi.ports["o2"])

    # Separación entre o3 y o2 
    sep_puertos = abs(mmi.ports["o2"].dy - mmi.ports["o3"].dy) 

    
    recto = c.add_ref(gf.components.straight(length=sep_puertos , cross_section=xs))
    recto.connect("o1", bend_der_1.ports["o2"])
    
    gf.routing.route_single(component=c,port1=recto.ports["o2"],port2=bend_izq_1.ports["o2"],cross_section=xs)

    c.add_port(name="o1", port=mmi.ports["o1"])
    c.add_port(name="o2", port=mmi.ports["o2"])
    c.add_port(name="o3", port=mmi.ports["o3"])
    c.add_port(name="o4", port=mmi.ports["o4"])
   
    c.draw_ports()
    return c


chip_ring_resonator = gf.Component()

ring1 = chip_ring_resonator.add_ref(ring_resonator_r( radio   = 35.0))
ring2 = chip_ring_resonator.add_ref(ring_resonator_r())
ring2.dmovey(80)

chip_ring_resonator.plot()
```

```python
#Create a cell component for an all-pass ring resonator, using 2x2 50:50 MMIs, with extra length parameter for different perimeters

gf.clear_cache()

@gf.cell
def ring_resonator_cell(
    W_MMI   = 6.6,
    L_MMI   = 34.2180,
    W_wg    = 1.0,
    L_wg    = 10.0,
    dy      = 0.1,
    radio   = 20.0, 
    L_extra = 0.0,
):
    c  = gf.Component()
    xs = gf.cross_section.strip(width=W_wg, layer="WG")
    
    # MMI 
    mmi = c.add_ref(mmi2x2(W_MMI=W_MMI, L_MMI=L_MMI, W_wg=W_wg, L_wg=L_wg, dy=dy))

    # Lado derecho
    bend_der_1 = c.add_ref(gf.components.bend_euler(angle=180, radius=radio + L_extra, cross_section=xs))
    bend_der_1.connect("o1", mmi.ports["o3"])

    # Lado izquierdo
    bend_izq_1 = c.add_ref(gf.components.bend_euler(angle=-180, radius=radio + L_extra, cross_section=xs))
    bend_izq_1.connect("o1", mmi.ports["o2"])

    sep_puertos = abs(mmi.ports["o2"].dy - mmi.ports["o3"].dy) 

    # Recto 
    recto = c.add_ref(gf.components.straight(length=sep_puertos , cross_section=xs))
    recto.connect("o1", bend_der_1.ports["o2"])
    
    gf.routing.route_single(component=c,port1=recto.ports["o2"],port2=bend_izq_1.ports["o2"],cross_section=xs)

    # Puertos 
    c.add_port(name="o1", port=mmi.ports["o1"])
    c.add_port(name="o2", port=mmi.ports["o2"])
    c.add_port(name="o3", port=mmi.ports["o3"])
    c.add_port(name="o4", port=mmi.ports["o4"])
   
    c.draw_ports()
    return c


chip_ring_resonator = gf.Component()

ring1 = chip_ring_resonator.add_ref(ring_resonator_cell(L_extra  = 20.0))
ring2 = chip_ring_resonator.add_ref(ring_resonator_cell(L_extra  = 50.0))
ring2.dmovey(120)

chip_ring_resonator.plot()
```

```python
gf.clear_cache()

sp = 250
border = 250
n=37

@gf.cell
def die_WL(dieL = 10000, dieW = 5000, border = 250, layer_box = "FLOORPLAN"):
    # Die specifications (Chip)
    box = gf.Component()
    obox = box.add_ref(gf.components.rectangle(size=(dieW,dieL),layer=layer_box))
    ibox = box.add_ref(gf.components.rectangle(size=(dieW-border*2,dieL-border*2),layer=layer_box)).dmovex(border).dmovey(border)
    box = gf.boolean(A=obox, B=ibox, operation="A-B", layer=layer_box)
    # Adding ports to a component 
    box.add_port(name="block@org", center=[border,border], width=1, orientation=0, layer=layer_box)
    box.draw_ports()

    y_start = 500
    for i in range(n):
        y_pos = y_start + i * sp

      
        in_arr = box.add_ref(
            gf.components.straight(length=border, cross_section=xs)
        ).dmovex(0).dmovey(y_pos)
        box.add_port(name=f"io_l_{i+1}", port=in_arr['o2'])

      
        out_arr = box.add_ref(
            gf.components.straight(length=border, cross_section=xs)
        ).drotate(180).dmovex(dieW).dmovey(y_pos)
        box.add_port(name=f"io_r_{i+1}", port=out_arr['o2'])

    
    return box

wafer = gf.Component()
dieW = 5000

c1 = wafer.add_ref(die_WL(dieW = dieW))

wafer.show()
wafer.plot()
```

```python
#Floorplaning and die layout

#Make an instance of the die as host component for your layout
#Add 3 sets of 3 straight waveguides, from left to right of the die, top, middle and bottom of the die
#Add 2 of each of the components above, with different lengths
#Connect all your components to the I/Os

layout = gf.Component()
die_WG = layout.add_ref(die_WL(dieL = 10000, dieW = 5000, border = 250, layer_box = "FLOORPLAN"))

rows = {
    "bottom": [1,  2,  3 ],
    "middle": [19, 20, 21],
    "top"   : [35, 36, 37],
}

def add_straight_row(parent, io_indices):
    for i in io_indices:
        gf.routing.route_single(
            parent,
            port1         = die_WG.ports[f'io_l_{i}'],
            port2         = die_WG.ports[f'io_r_{i}'],
            cross_section = xs,
            allow_width_mismatch=True,
        )

add_straight_row(layout, rows["bottom"])
add_straight_row(layout, rows["middle"])
add_straight_row(layout, rows["top"])

layout.plot()
layout.show()
```
